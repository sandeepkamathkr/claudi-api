```python
# Load environment variables from .env file
from dotenv import load_dotenv
load_dotenv()
```

```python
# Create an API client
from anthropic import Anthropic

client = Anthropic()
# Using Claude 3.5 Sonnet
model = "claude-sonnet-5"
```

```python

# add a message to the conversation
def add_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model=model,
        messages=messages,
        max_tokens=1000
    )
    return message.content[0].text
```

```python
# Make a starting list of messages
messages = []

# Add in the initial user question of "Define quantum computing in one sentence."
add_message(messages, "Define quantum computing in one sentence.")


# Pass the list of messages into 'chat' to get an answer
response = chat(messages)

# Take the answer and add it as an assistant message to the list of messages
add_assistant_message(messages, response)

# Add in the user's follow-up question of "Write another sentence"
add_message(messages, "Write another sentence")

# Pass the list of messages into 'chat' to get an answer
response = chat(messages)
response
```
