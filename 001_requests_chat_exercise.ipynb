```python
# Load environment variables from .env file
from dotenv import load_dotenv
load_dotenv()
```

```python
# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-4-5"
```

```python

# add a message to the conversation
def add_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model=model,
        messages=messages,
        max_tokens=1000
    )
    text_blocks = [block.text for block in message.content if block.type == "text"]
    if not text_blocks:
        return ""  # or raise a meaningful error
    return text_blocks[0]
```

```python
# Make a starting list of messages
MAX_MESSAGES = 10  # keep last 5 turns (user + assistant = 2 per turn)
messages = []

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Goodbye!")
        break

    add_message(messages, user_input)

    # Trim to last MAX_MESSAGES before sending
    trimmed = messages[-MAX_MESSAGES:]
    response = chat(trimmed)

    print("--------------------")
    print("Claude:", response)
    print("--------------------")

    add_assistant_message(messages, response)
```
